```python
# EXERCISE 2
# ReACT Code Generation
# Tools Used: ChatGPT + Google Colab


# --------------------------------------------------
# PLAN
# --------------------------------------------------

print("""
PLAN:

1. Create a function that accepts prices and a discount.
2. Validate the inputs before doing calculations.
3. Add the prices together.
4. Apply the discount.
5. Round the result to two decimal places.
6. Test normal inputs and invalid inputs.
""")


# --------------------------------------------------
# FIRST GENERATED VERSION
# --------------------------------------------------

def calculate_discounted_total(prices, discount_percent):
    """
    Calculates the total price after applying a discount.
    """

    if not isinstance(prices, list) or len(prices) == 0:
        raise ValueError("Prices must be a non-empty list.")

    for price in prices:
        if not isinstance(price, (int, float)):
            raise ValueError("Every price must be numeric.")

        if price < 0:
            raise ValueError("Prices cannot be negative.")

    if not isinstance(discount_percent, (int, float)):
        raise ValueError("Discount must be numeric.")

    if discount_percent < 0 or discount_percent > 100:
        raise ValueError("Discount must be between 0 and 100.")

    subtotal = sum(prices)

    discount_amount = subtotal * (discount_percent / 100)

    final_total = subtotal - discount_amount

    return round(final_total, 2)


# --------------------------------------------------
# RUN / TEST
# --------------------------------------------------

print("\nFIRST TEST RUN")
print("-" * 40)

print("Normal purchase:")
print(calculate_discounted_total([10, 20, 30], 10))

print("\nDecimal prices:")
print(calculate_discounted_total([19.99, 5.01], 20))


try:
    calculate_discounted_total([], 10)
except ValueError as error:
    print("\nEmpty list test passed:", error)


try:
    calculate_discounted_total([10, -5], 10)
except ValueError as error:
    print("Negative price test passed:", error)


try:
    calculate_discounted_total([10, 20], 150)
except ValueError as error:
    print("Invalid discount test passed:", error)


# --------------------------------------------------
# OBSERVATION
# --------------------------------------------------

print("\n" + "=" * 60)
print("OBSERVATION")
print("=" * 60)

print("""
The basic tests worked correctly.

However, there is another edge case.

In Python, True and False can behave like the numbers 1 and 0.
Because the first version checks whether a value is an int or float,
it may accidentally accept Boolean values.

The code should be improved to reject Boolean values.
""")


# --------------------------------------------------
# REVISION / FIX
# --------------------------------------------------

print("=" * 60)
print("REVISION PROMPT")
print("=" * 60)

revision_prompt = """
Revise the Python function after reviewing the test results.

The first version works for normal values, empty lists,
negative prices, and invalid discounts.

However, Python may treat True and False as integers.

Improve the program so that:

1. True and False cannot be used as product prices.
2. True and False cannot be used as the discount.
3. All previous validation still works.
4. The final result is still rounded to two decimal places.
5. ValueError is raised for invalid inputs.

Return the improved Python function.
"""

print(revision_prompt)


# --------------------------------------------------
# IMPROVED VERSION
# --------------------------------------------------

def calculate_discounted_total_improved(prices, discount_percent):
    """
    Calculates the total price after applying a discount
    while validating all inputs.
    """

    if not isinstance(prices, list) or len(prices) == 0:
        raise ValueError("Prices must be a non-empty list.")

    for price in prices:

        if isinstance(price, bool):
            raise ValueError("Boolean values cannot be used as prices.")

        if not isinstance(price, (int, float)):
            raise ValueError("Every price must be numeric.")

        if price < 0:
            raise ValueError("Prices cannot be negative.")

    if isinstance(discount_percent, bool):
        raise ValueError("Boolean values cannot be used as discounts.")

    if not isinstance(discount_percent, (int, float)):
        raise ValueError("Discount must be numeric.")

    if discount_percent < 0 or discount_percent > 100:
        raise ValueError("Discount must be between 0 and 100.")

    subtotal = sum(prices)

    discount_amount = subtotal * (discount_percent / 100)

    final_total = subtotal - discount_amount

    return round(final_total, 2)


# --------------------------------------------------
# RUN AGAIN
# --------------------------------------------------

print("\n" + "=" * 60)
print("FINAL TEST RESULTS")
print("=" * 60)


print(
    "Normal purchase:",
    calculate_discounted_total_improved([10, 20, 30], 10)
)


print(
    "Decimal prices:",
    calculate_discounted_total_improved([19.99, 5.01], 20)
)


try:
    calculate_discounted_total_improved([], 10)
except ValueError:
    print("PASS: Empty list rejected")


try:
    calculate_discounted_total_improved([10, -5], 10)
except ValueError:
    print("PASS: Negative price rejected")


try:
    calculate_discounted_total_improved([10, 20], 150)
except ValueError:
    print("PASS: Invalid discount rejected")


try:
    calculate_discounted_total_improved([True, 10], 10)
except ValueError:
    print("PASS: Boolean price rejected")


try:
    calculate_discounted_total_improved([10, 20], True)
except ValueError:
    print("PASS: Boolean discount rejected")


print("\n" + "=" * 60)
print("ReACT PROCESS COMPLETED")
print("PLAN -> CODE -> RUN -> OBSERVE -> FIX -> RUN AGAIN")
print("=" * 60)
```
